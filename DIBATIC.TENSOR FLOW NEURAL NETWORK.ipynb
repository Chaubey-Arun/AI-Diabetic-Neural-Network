```python
import pandas as pd
import numpy as np

# Sample data for df_students
data = {'Name': [False] * 22,
        'StudyHours': [False] * 22,
        'Grade': [False] * 22}

df_students = pd.DataFrame(data)

# Function to randomly insert NaN values
def add_null_values(df, num_nulls):
    for _ in range(num_nulls):
        # Randomly choose a row and column to insert NaN
        row = np.random.randint(0, df.shape[0])  # Random row
        col = np.random.choice(df.columns)      # Random column
        df.loc[row, col] = np.nan  # Set the value to NaN
    return df

# Add 5 null values to the DataFrame
df_students = add_null_values(df_students, 5)

# Show the updated DataFrame
print(df_students)
```

```python
import pandas as pd

# Assuming 'dia' is already defined and loaded with the data.

# Define the diabetic classes
diabetic_classes = ['Non Diabetic', 'Diabetic']

# Print the first five columns of the DataFrame
print(dia.columns[0:5].values, 'Diabetic')

# Sample 10 rows and iterate over them
for index, row in dia.sample(10).iterrows():
    # Print the row values, using column names for clarity
    print('[', row['PatientID'], row['Pregnancies'], row['PlasmaGlucose'], row['DiastolicBloodPressure'], 
          row['TricepsThickness'], int(row['SerumInsulin']), ']', diabetic_classes[int(row['Diabetic'])])
```

```python
from sklearn.model_selection import train_test_split
features=  ['Pregnancies', 'PlasmaGlucose', 'DiastolicBloodPressure', 'TricepsThickness']
label = 'Diabetic'
#split data 70%-30% into traning
x_train ,x_test,y_train,y_test1 = train_test_split(dia[features].values,
                                                  dia[label].values,
                                                  test_size=0.30,
                                                  random_state=0)
print('Traning set:%d, Test set : %d \n'%(len(x_train),len(x_test)))
print('sample of features and labels;')

for n in range(0,24):
    print(x_train[n],y_train[n],'('+diabetic_classes[y_train[n]]+')')
```

```python
import tensorflow 
from tensorflow import keras
from tensorflow.keras import models
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils
from tensorflow.keras import optimizers

#set random seed for reproducabiltiy 
tensorflow.random.set_seed(0)

print('Libraries imported.')
print('Tensorflow version :',tensorflow.__version__)
```

```python
x_train = x_train.astype('float32')
x_test = x_test.astype('float32')

#set data types for categorical labels 
y_train= utils.to_categorical(y_train)
y_test1=utils.to_categorical(y_test1)
print('Ready..2.')
```


Define a neural network

```python
hl =10 #Number of Hidden layer nodes
model= Sequential()
model.add(Dense(hl,input_dim=len(features),activation='relu'))
model.add(Dense(hl,input_dim=hl,activation='relu'))
model.add(Dense(len(diabetic_classes),input_dim=hl,activation ='softmax'))

print(model.summary())
```

Train the model


```python
#hyper-parameters for optimizer 
learning_rate = 0.00001
#opt = optimizer.Adam (lr=Learnign rate)
opt= optimizers.Adam(learning_rate=0.00001)

model.compile(loss='categorical_crossentropy',
              optimizer=opt,
              metrics=['accuracy'])
num_epochs= 50 
history = model.fit(x_train,y_train,epochs=num_epochs,batch_size=10,validation_data=(x_test,y_test1))
```
